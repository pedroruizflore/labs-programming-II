```python
def outer():
  x = 10
  def inner():
    return x
  return inner

f = outer()
print(f())
```

```python
def deco(fn):
  def wrapper():
    print("before")
    fn()
    print("after")
  return wrapper

@deco
def hello():
  print("hello")

hello()
```

```python
class MyRange:
  def __iter__(self):
    return self
  def __next__(self):
    raise StopIteration

for x in MyRange():
  print(x)
```
