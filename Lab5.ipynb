```python
from abc import ABC , abstractmethod


class Vehicle ( ABC ) :
  @abstractmethod
  def calculate_fare ( self , distance ) :
    pass

  @abstractmethod
  def get_vehicle_type ( self ) :
    pass


class UberX ( Vehicle ) :

  def calculate_fare ( self , distance ) :
    return distance * 10

  def get_vehicle_type ( self ) :
    return " UberX "

uberx = UberX ()
print ( uberx . calculate_fare (5) )
```

Student Questions
1. Why can't we instantiate Vehicle? Python prevents this because Vehicle is an Abstract Base Class (ABC) containing @abstractmethod. It serves as a blueprint or "contract" rather than a complete object.

2. What happens if UberX does not implement get vehicle type()? get_vehicle_type()? Python will raise a TypeError and refuse to instantiate UberX. A subclass must implement all abstract methods defined in its parent ABC to be considered "concrete."

3. Is ABC enforcing structure at runtime or compile-time? In Python, this is enforced at runtime when you attempt to create an instance of the class.

4. Why is this useful for large companies like Uber? It ensures consistency. Developers can create hundreds of new vehicle types, and the system is guaranteed that every single one will have a calculate_fare method, preventing system crashes.

```python
class UberBike ( Vehicle ) :

  def calculate_fare ( self , distance ) :
    return distance * 5

  def get_vehicle_type ( self ) :
    return " UberBike "

uberx = UberX ()
bike = UberBike ()

print ( isinstance ( uberx , Vehicle ) )
print ( isinstance ( bike , Vehicle ) )
```

Student Questions
1. Is UberBike automatically a Vehicle? Yes, because it inherits from the Vehicle class

2. What happens if we remove (Vehicle) from UberBike? It would become a standalone class. It would lose its formal "IS-A" relationship, and isinstance would return False.

3. Is inheritance modeling IS-A or HAS-A? Inheritance models the IS-A relationship.

```python
vehicles = [ UberX () , UberBike () ]

for vehicle in vehicles :
  print ( vehicle . get_vehicle_type () ,
         vehicle . calculate_fare (10) )
```

Student Questions
1. Where does runtime polymorphism occur? It occurs inside the loop where vehicle.calculate_fare(10) is called.

2. What would happen if method names differed? The loop would fail with an AttributeError because it expects a uniform method name across all objects in the list.

3. Why is this better than using if-else checks? It makes the code "extensible". You can add a new vehicle type to the vehicles list without ever changing the loop's code.

```python
class ElectricMixin :

  def charge_battery ( self ) :
    return " Charging battery ... "


class UberGreen ( ElectricMixin , UberX ) :
  pass


green = UberGreen ()

print ( green . get_vehicle_type () )
print ( green . charge_battery () )
```

Student Questions
1. Why is ElectricMixin listed first? In Python's Multiple Inheritance, the order determines which class is searched first for a method.

2. Is ElectricMixin modeling IS-A? No. Mixins usually model a capability rather than a core identity.

3. What happens if both parents define same method? Python will use the method from the class listed first in the inheritance parentheses.


```python
class Driver :

  def __init__ (self , name , vehicle):
    self.name = name
    self.vehicle = vehicle

  def start_trip (self , distance) :
    fare = self.vehicle.calculate_fare(distance)
    return f" {self.name} driving {self.vehicle.get_vehicle_type()} - Fare: {fare}"


driver1 = Driver ( " Alice " , UberX () )
driver2 = Driver ( " Bob " , UberBike () )

print(driver1.start_trip(10))
print(driver2.start_trip(10))
```

Student Questions
1. Is Driver inheriting from Vehicle? No. A Driver is not a type of Vehicle.

2. What relationship is this modeling? It models Composition (HAS-A)

3. Why is composition flexible? It allows a Driver to switch vehicles easily at runtime just by changing the self.vehicle attribute.

```python
class UberApp :

  def request_ride ( self , driver , distance ) :
    return driver . start_trip ( distance )

app = UberApp ()

print ( app . request_ride ( driver1 , 5) )
```

Student Questions
1. How is dependency different from composition? In composition, the Driver "owns" the Vehicle as a permanent attribute. In dependency, the UberApp only "uses" the Driver temporarily as a parameter in a method.

2. Does UberApp permanently store Driver? No. It receives the driver object, performs a task, and then the relationship ends.

3. Why is loose coupling important in real systems? It allows parts of the system to change independently. The UberApp doesn't need to know how Driver or Vehicle works; it just needs to know it can call request_ride.
